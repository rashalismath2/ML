```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import signal
from sklearn.decomposition import FastICA
```

```python
t = np.linspace(0, 8, 2000)

s1 = np.sin(2*t)
s2 = signal.square(3*t)

S = np.c_[s1, s2]
```

```python
plt.plot(s1)
plt.title("Source 1")
plt.show()

plt.plot(s2)
plt.title("Source 2")
plt.show()

plt.plot(S)
plt.title("Source")
plt.show()
```

```python
A = np.array([
    [1.0, 0.5],
    [0.5, 1.0]
])

X = S @ A.T
```

```python
plt.plot(X[:,0])
plt.title("Mixture 1")
plt.show()

plt.plot(X[:,1])
plt.title("Mixture 2")
plt.show()
```

```python
ica = FastICA(n_components=2)
S_recovered = ica.fit_transform(X)
```

```python
plt.plot(S_recovered[:,0])
plt.title("Recovered Source 1")
plt.show()
```

```python
plt.plot(S_recovered[:,1])
plt.title("Recovered Source 2")
plt.show()
```
